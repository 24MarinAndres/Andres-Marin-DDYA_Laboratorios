## 1) Descripción del problema **[1 pts]**
Ejercicio: Recorrido de una red de distribución
Una empresa de mensajería tiene varios centros de distribución conectados mediante rutas terrestres.

Cada centro será representado mediante un vértice y cada carretera que conecta dos centros será representada mediante una arista.

La empresa desea desarrollar un programa que permita recorrer la red utilizando los algoritmos:

BFS (Breadth-First Search).

DFS (Depth-First Search).

Red de distribución
La empresa tiene los siguientes centros:

A - Centro principal B - Centro Norte C - Centro Sur D - Centro Occidente E - Centro Oriente F - Centro Industrial G - Centro Aeropuerto
Las conexiones entre ellos son:

A - B A - C B - D B - E C - F E - G F - G
Gráficamente, la red puede representarse así:

        A           /   \          B     C         / \     \        D   E     F             \   /               G

Punto 2. Recorrido BFS
El administrador necesita identificar primero los centros que se encuentran más cerca del centro principal en cantidad de conexiones.

Implemente el algoritmo BFS comenzando desde el centro A.

A -> B -> C -> D -> E -> F -> G

Puede implementarlo mediante recursividad o utilizando una pila.

Un posible recorrido sería:

A -> B -> D -> E -> G -> F -> C
El resultado puede variar dependiendo del orden en que se almacenen los vecinos.

Punto 5. Buscar un centro utilizando BFS

## 6) Código funcional 

```python
import matplotlib.pyplot as plt
from collections import deque
from copy import deepcopy

GRAFO_BASE = {
    "A": ["B", "C"],
    "B": ["A", "D", "E"],
    "C": ["A", "F"],
    "D": ["B"],
    "E": ["B", "G"],
    "F": ["C", "G"],
    "G": ["E", "F"]
}

grafo = deepcopy(GRAFO_BASE)

def mostrar_lista_adyacencia(grafo):
    print("\nLISTA DE ADYACENCIA")
    print("-" * 45)
    for centro, vecinos in grafo.items():
        print(f"{centro}: {vecinos}")

def dibujar_grafo(grafo):
    # ÚNICO diagrama visual del ejercicio: solo Punto 1
    posiciones = {
        "A": (0, 3),
        "B": (-1.7, 1.8),
        "C": (1.7, 1.8),
        "D": (-2.7, 0.4),
        "E": (-0.8, 0.4),
        "F": (1.7, 0.4),
        "G": (0.5, -1.0)
    }

    plt.figure(figsize=(9, 6))
    aristas_dibujadas = set()

    for origen in grafo:
        for destino in grafo[origen]:
            arista = tuple(sorted((origen, destino)))
            if arista not in aristas_dibujadas:
                x1, y1 = posiciones[origen]
                x2, y2 = posiciones[destino]
                plt.plot([x1, x2], [y1, y2], linewidth=2, zorder=1)
                aristas_dibujadas.add(arista)

    for nodo in grafo:
        x, y = posiciones[nodo]
        plt.text(
            x, y, nodo,
            fontsize=14,
            ha="center",
            va="center",
            zorder=3,
            bbox={"boxstyle": "circle,pad=0.55", "linewidth": 2}
        )

    plt.title("Red de distribución — Grafo no dirigido", fontsize=15)
    plt.axis("equal")
    plt.axis("off")
    plt.tight_layout()
    plt.show()

def punto_1(grafo):
    mostrar_lista_adyacencia(grafo)
    dibujar_grafo(grafo)

def bfs(grafo, inicio):
    if inicio not in grafo:
        raise ValueError("El centro inicial no existe.")

    cola = deque([inicio])
    visitados = {inicio}
    recorrido = []
    niveles = {inicio: 0}

    while cola:
        actual = cola.popleft()
        recorrido.append(actual)

        for vecino in grafo[actual]:
            if vecino not in visitados:
                visitados.add(vecino)
                niveles[vecino] = niveles[actual] + 1
                cola.append(vecino)

    return recorrido, niveles

def mostrar_bfs(grafo, inicio):
    try:
        recorrido, niveles = bfs(grafo, inicio)
    except ValueError as error:
        print("Error:", error)
        return None

    print(f"\nRECORRIDO BFS DESDE {inicio}")
    print("-" * 45)
    print(" -> ".join(recorrido))
    print("\nNIVELES")

    for nivel in range(max(niveles.values()) + 1):
        nodos = [nodo for nodo in recorrido if niveles[nodo] == nivel]
        print(f"Nivel {nivel}:", ", ".join(nodos))

    return recorrido, niveles

def dfs_recursivo(grafo, actual, visitados, recorrido):
    visitados.add(actual)
    recorrido.append(actual)

    for vecino in grafo[actual]:
        if vecino not in visitados:
            dfs_recursivo(grafo, vecino, visitados, recorrido)

def dfs(grafo, inicio):
    if inicio not in grafo:
        raise ValueError("El centro inicial no existe.")

    visitados = set()
    recorrido = []
    dfs_recursivo(grafo, inicio, visitados, recorrido)
    return recorrido

def mostrar_dfs(grafo, inicio):
    try:
        recorrido = dfs(grafo, inicio)
    except ValueError as error:
        print("Error:", error)
        return None

    print(f"\nRECORRIDO DFS DESDE {inicio}")
    print("-" * 45)
    print(" -> ".join(recorrido))
    return recorrido

def punto_4(grafo):
    print("\nCENTROS DISPONIBLES")
    print(", ".join(grafo.keys()))

    inicio = input("Ingrese el centro inicial: ").strip().upper()

    if inicio not in grafo:
        print("Error: el centro ingresado no existe.")
        return None

    mostrar_bfs(grafo, inicio)
    mostrar_dfs(grafo, inicio)
    return inicio

def buscar_centro_bfs(grafo, inicio, destino):
    if inicio not in grafo:
        raise ValueError("El centro inicial no existe.")
    if destino not in grafo:
        raise ValueError("El centro destino no existe.")

    cola = deque([inicio])
    visitados = {inicio}
    padre = {inicio: None}

    while cola:
        actual = cola.popleft()

        if actual == destino:
            break

        for vecino in grafo[actual]:
            if vecino not in visitados:
                visitados.add(vecino)
                padre[vecino] = actual
                cola.append(vecino)

    if destino not in padre:
        return None

    ruta = []
    actual = destino

    while actual is not None:
        ruta.append(actual)
        actual = padre[actual]

    ruta.reverse()
    return ruta

def punto_5(grafo):
    print("\nCENTROS DISPONIBLES")
    print(", ".join(grafo.keys()))

    inicio = input("Centro inicial: ").strip().upper()
    destino = input("Centro que desea buscar: ").strip().upper()

    try:
        ruta = buscar_centro_bfs(grafo, inicio, destino)
    except ValueError as error:
        print("Error:", error)
        return None

    if ruta is None:
        print("No existe una ruta hasta el centro solicitado.")
        return None

    print("\nCentro encontrado.")
    print("Ruta:", " -> ".join(ruta))
    print("Cantidad de conexiones:", len(ruta) - 1)
    return ruta

def agregar_centro(grafo, nuevo, conexiones):
    nuevo = nuevo.strip().upper()

    if nuevo == "":
        raise ValueError("El nombre del centro no puede estar vacío.")
    if nuevo in grafo:
        raise ValueError("El centro ya existe.")

    conexiones_limpias = []

    for centro in conexiones:
        centro = centro.strip().upper()

        if centro == nuevo:
            raise ValueError("Un centro no puede conectarse consigo mismo.")
        if centro not in grafo:
            raise ValueError(f"El centro {centro} no existe.")
        if centro not in conexiones_limpias:
            conexiones_limpias.append(centro)

    grafo[nuevo] = []

    for centro in conexiones_limpias:
        grafo[nuevo].append(centro)
        grafo[centro].append(nuevo)

    return grafo

def punto_6(grafo):
    nuevo = input("\nIngrese el nuevo centro: ").strip().upper()

    print("Centros existentes:", ", ".join(grafo.keys()))

    conexiones_texto = input(
        "Conectar con qué centros (separados por coma, ejemplo D,G): "
    )

    conexiones = [
        centro.strip().upper()
        for centro in conexiones_texto.split(",")
        if centro.strip()
    ]

    try:
        agregar_centro(grafo, nuevo, conexiones)
    except ValueError as error:
        print("Error:", error)
        return None

    print(f"\nCentro {nuevo} agregado correctamente.")

    # Punto 6.1: mostrar lista actualizada
    mostrar_lista_adyacencia(grafo)

    # No se dibuja otro diagrama porque el enunciado no lo exige.

    # Punto 6.4: seleccionar de nuevo el vértice inicial
    inicio = input(
        "\nSeleccione nuevamente el vértice inicial: "
    ).strip().upper()

    if inicio not in grafo:
        print("Error: el centro ingresado no existe.")
        return nuevo

    # Punto 6.2 y 6.3
    mostrar_bfs(grafo, inicio)
    mostrar_dfs(grafo, inicio)

    return nuevo

def main():
    while True:
        print("\n" + "=" * 55)
        print("RED DE DISTRIBUCIÓN")
        print("=" * 55)
        print("1. Representar el grafo")
        print("2. Recorrido BFS desde A")
        print("3. Recorrido DFS desde A")
        print("4. Seleccionar el punto inicial")
        print("5. Buscar un centro utilizando BFS")
        print("6. Agregar nuevos centros y conexiones")
        print("7. Salir")

        opcion = input("\nSeleccione una opción: ").strip()

        if opcion == "1":
            punto_1(grafo)
        elif opcion == "2":
            mostrar_bfs(grafo, "A")
        elif opcion == "3":
            mostrar_dfs(grafo, "A")
        elif opcion == "4":
            punto_4(grafo)
        elif opcion == "5":
            punto_5(grafo)
        elif opcion == "6":
            punto_6(grafo)
        elif opcion == "7":
            print("Programa finalizado.")
            break
        else:
            print("Opción inválida.")

print("Programa cargado correctamente.")

main()
```

```python
# ============================================================
# CELDA 2 — PRUEBAS CON assert
# ============================================================

from copy import deepcopy

g = deepcopy(GRAFO_BASE)

# 1. Vértices originales
assert set(g.keys()) == {"A", "B", "C", "D", "E", "F", "G"}

# 2. Lista de adyacencia completa
assert g["A"] == ["B", "C"]
assert g["B"] == ["A", "D", "E"]
assert g["C"] == ["A", "F"]
assert g["D"] == ["B"]
assert g["E"] == ["B", "G"]
assert g["F"] == ["C", "G"]
assert g["G"] == ["E", "F"]

# 3. El grafo es no dirigido
for nodo in g:
    for vecino in g[nodo]:
        assert nodo in g[vecino]

# 4. BFS desde A
recorrido_bfs, niveles = bfs(g, "A")

assert recorrido_bfs == ["A", "B", "C", "D", "E", "F", "G"]

# 5. Niveles BFS
assert niveles == {
    "A": 0,
    "B": 1,
    "C": 1,
    "D": 2,
    "E": 2,
    "F": 2,
    "G": 3
}

# 6. BFS visita cada vértice una vez
assert len(recorrido_bfs) == len(set(recorrido_bfs))
assert len(recorrido_bfs) == len(g)

# 7. DFS desde A
recorrido_dfs = dfs(g, "A")

assert recorrido_dfs == ["A", "B", "D", "E", "G", "F", "C"]

# 8. DFS visita cada vértice una vez
assert len(recorrido_dfs) == len(set(recorrido_dfs))
assert len(recorrido_dfs) == len(g)

# 9. BFS puede empezar desde B
bfs_b, niveles_b = bfs(g, "B")
assert bfs_b[0] == "B"
assert set(bfs_b) == set(g.keys())

# 10. DFS puede empezar desde B
dfs_b = dfs(g, "B")
assert dfs_b[0] == "B"
assert set(dfs_b) == set(g.keys())

# 11. Buscar F desde A
ruta = buscar_centro_bfs(g, "A", "F")
assert ruta == ["A", "C", "F"]

# 12. Cantidad de conexiones
assert len(ruta) - 1 == 2

# 13. Buscar un nodo desde sí mismo
assert buscar_centro_bfs(g, "A", "A") == ["A"]

# 14. Centro inalcanzable
grafo_desconectado = deepcopy(GRAFO_BASE)
grafo_desconectado["H"] = []

assert buscar_centro_bfs(
    grafo_desconectado,
    "A",
    "H"
) is None

# 15. BFS con nodo inexistente
error_bfs = False
try:
    bfs(g, "Z")
except ValueError:
    error_bfs = True

assert error_bfs is True

# 16. DFS con nodo inexistente
error_dfs = False
try:
    dfs(g, "Z")
except ValueError:
    error_dfs = True

assert error_dfs is True

# 17. Destino inexistente
error_destino = False
try:
    buscar_centro_bfs(g, "A", "Z")
except ValueError:
    error_destino = True

assert error_destino is True

# 18. Agregar H conectado con D y G
grafo_ampliado = deepcopy(GRAFO_BASE)

agregar_centro(
    grafo_ampliado,
    "H",
    ["D", "G"]
)

assert "H" in grafo_ampliado
assert grafo_ampliado["H"] == ["D", "G"]

# 19. Conexiones en ambos sentidos
assert "H" in grafo_ampliado["D"]
assert "H" in grafo_ampliado["G"]

# 20. BFS después de agregar H
bfs_h, niveles_h = bfs(grafo_ampliado, "A")
assert "H" in bfs_h
assert len(bfs_h) == 8

# 21. DFS después de agregar H
dfs_h = dfs(grafo_ampliado, "A")
assert "H" in dfs_h
assert len(dfs_h) == 8

# 22. Evitar centro duplicado
error_duplicado = False
try:
    agregar_centro(grafo_ampliado, "H", ["A"])
except ValueError:
    error_duplicado = True

assert error_duplicado is True

# 23. Evitar conexión con nodo inexistente
error_conexion = False
try:
    temporal = deepcopy(GRAFO_BASE)
    agregar_centro(temporal, "H", ["Z"])
except ValueError:
    error_conexion = True

assert error_conexion is True

# 24. Evitar lazo H-H
error_lazo = False
try:
    temporal = deepcopy(GRAFO_BASE)
    agregar_centro(temporal, "H", ["H"])
except ValueError:
    error_lazo = True

assert error_lazo is True

# 25. No duplicar conexiones repetidas
temporal = deepcopy(GRAFO_BASE)

agregar_centro(
    temporal,
    "H",
    ["D", "D", "G", "G"]
)

assert temporal["H"] == ["D", "G"]
assert temporal["D"].count("H") == 1
assert temporal["G"].count("H") == 1

print("\n" + "=" * 55)
print("TODAS LAS PRUEBAS PASARON CORRECTAMENTE")
print("=" * 55)

```
