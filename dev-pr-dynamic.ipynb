```python
import matplotlib.pyplot as plt

n = list(range(1,11))

mejor = [1]*10
peor = n

plt.plot(n, mejor, label ="mejor caso =0(1)")
plt.plot(n, peor, label ="peor caso =0(n)")

plt.xlabel("tamaño de entrada n")
plt.ylabel("operaciones")
plt.title("analisis de complejidad")
plt.legend()
plt.grid()
plt.show()
```

```python
def plantas(n, memo):

    if n in memo:
        return memo[n]
    if n == 1:
        return 1
    if n == 0:
        return 0

    memo[n]= plantas (n-1, memo)+ plantas(n-2,memo)

    return memo[n]

def main():
    memo ={}
    n = int(input("digite catidad de plantas a evaluar: "))

    resultado = plantas(n, memo)
    print("numero de plantas producidas en el mes", n ,":", resultado)
    print("memo", memo)


main()
```
