## 1) Descripción del problema **[1 pts]**

Una tienda registra los códigos numéricos de sus productos a medida que estos llegan. Debido a esto, los códigos quedan inicialmente almacenados en un orden cualquiera. El problema consiste en construir un programa que permita ingresar una cantidad **n** de productos, guardar sus códigos, ordenarlos de menor a mayor y posteriormente buscar un código específico.

Para resolver el problema se utilizan dos algoritmos principales. **Merge Sort** se encarga del ordenamiento mediante la estrategia de dividir la lista en partes más pequeñas, ordenar cada parte y combinarlas nuevamente. Después de tener la lista ordenada, se utiliza **Búsqueda Binaria**, que compara el código buscado con el elemento central y descarta la mitad de la lista en cada paso.

## 7) Análisis de complejidad **[15 pts]**

### Merge Sort

Merge Sort divide la lista en dos mitades y después combina los resultados recorriendo los elementos. Su recurrencia es:

\[
T(n)=2T(n/2)+n
\]

Al expandirla:

\[
T(n)=2^kT(n/2^k)+kn
\]

Cuando \(n/2^k=1\), entonces \(2^k=n\) y \(k=\log_2(n)\). Por lo tanto:

\[
T(n)=nT(1)+n\log_2(n)=O(n\log n)
\]

La función `merge` recorre en conjunto todos los elementos de las dos mitades, por lo cual:

\[
T_{merge}(n)=O(n)
\]

### Búsqueda Binaria

En cada iteración se descarta aproximadamente la mitad de la lista:

\[
T(n)=T(n/2)+1
\]

Después de \(k\) divisiones:

\[
T(n)=T(n/2^k)+k
\]

Al llegar a un solo elemento, \(k=\log_2(n)\), de modo que el peor caso es:

\[
T(n)=O(\log n)
\]

El mejor caso ocurre cuando el código buscado está exactamente en la posición central de la lista en la primera comparación, por lo que es **O(1)**.

### Entrada de códigos

Para registrar los **n** productos se realiza un recorrido de tamaño \(n\):

\[
T(n)=\sum_{i=1}^{n}1=n=O(n)
\]

### Complejidad total

\[
T(n)=O(n)+O(n\log n)+O(\log n)
\]

El término dominante es \(n\log n\), entonces:

\[
\boxed{T(n)=O(n\log n)}
\]

| Parte del programa | Mejor caso | Peor caso |
|---|---:|---:|
| Entrada de códigos | O(n) | O(n) |
| Merge Sort | O(n log n) | O(n log n) |
| Búsqueda Binaria | O(1) | O(log n) |
| Programa completo | O(n log n) | O(n log n) |

En el programa completo, el mejor y el peor caso tienen el mismo orden asintótico porque Merge Sort siempre realiza el proceso de división y combinación, incluso si la lista ya estaba ordenada.

### Gráfica de crecimiento



### Complejidad espacial

Merge Sort utiliza listas auxiliares de tamaño total proporcional a **n** y una pila recursiva de profundidad aproximada \(\log n\):

\[
O(n)+O(\log n)=O(n)
\]

Por tanto, la complejidad espacial total es **O(n)**.

> **Nota:** el análisis supone que cada dato válido se ingresa después de un número finito de intentos. Los ciclos de validación dependen de la interacción del usuario y no del tamaño de la lista.


## 8) Código funcional **[15 pts]**

La implementación se divide en cuatro partes principales:

- `merge_sort(lista)`: divide recursivamente la lista y ordena cada mitad.
- `merge(izq, der)`: combina dos listas ya ordenadas en una sola lista ordenada.
- `busqueda_binaria(lista, obj)`: busca el código en la lista ordenada y retorna `(True, posición)` si lo encuentra o `(False, -1)` si no existe.
- `main()`: controla la entrada de datos, las validaciones, el ordenamiento, la búsqueda y la salida del programa.

El código incluye validaciones para evitar que una entrada incorrecta termine la ejecución del programa.


```python
def merge_sort(lista):
    if len(lista) <= 1:
        return lista

    medio = len(lista) // 2
    izquierda = lista[:medio]
    derecha = lista[medio:]

    izquierda_ordenada = merge_sort(izquierda)
    derecha_ordenada = merge_sort(derecha)

    return merge(izquierda_ordenada, derecha_ordenada)


def merge(izq, der):
    res = []
    i = 0
    j = 0

    while i < len(izq) and j < len(der):
        if izq[i] <= der[j]:
            res.append(izq[i])
            i += 1
        else:
            res.append(der[j])
            j += 1

    res.extend(izq[i:])
    res.extend(der[j:])

    return res


def busqueda_binaria(lista, obj):
    izq = 0
    der = len(lista) - 1

    while izq <= der:
        medio = (izq + der) // 2

        if lista[medio] == obj:
            return True, medio
        elif lista[medio] < obj:
            izq = medio + 1
        else:
            der = medio - 1

    return False, -1


def main():
    # Registrar la cantidad de productos.
    while True:
        try:
            n = int(input("Cantidad de productos: "))
            if n > 0:
                break
            print("Debe ser mayor que 0. Vuelva a introducir un número apropiado.")
        except ValueError:
            print("Ingrese un número entero apropiado.")

    # Registrar los códigos de los productos.
    codigos = []

    for i in range(n):
        while True:
            try:
                cod = int(input("Código " + str(i + 1) + ": "))

                if cod >= 0:
                    codigos.append(cod)
                    break

                print("No puede ser negativo. Vuelva a introducir un número apropiado.")
            except ValueError:
                print("Ingrese un número entero apropiado.")

    # Ordenar los códigos con Merge Sort.
    print("Lista original:", codigos)
    ordenados = merge_sort(codigos)
    print("Lista ordenada:", ordenados)

    # Pedir y validar el código que se desea buscar.
    while True:
        try:
            buscar = int(input("Código a buscar: "))

            if buscar >= 0:
                break

            print("No puede ser negativo. Vuelva a introducir un número apropiado.")
        except ValueError:
            print("Ingrese un número entero apropiado.")

    # Realizar la búsqueda binaria sobre la lista ordenada.
    encontrado, pos = busqueda_binaria(ordenados, buscar)

    if encontrado:
        print("El producto con código", buscar, "fue encontrado.")
        print("Posición en la lista ordenada:", pos + 1)
    else:
        print("El producto con código", buscar, "no se encuentra registrado.")


main()
```
