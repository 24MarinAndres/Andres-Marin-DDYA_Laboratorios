```python
"""
Ejercicio: Organización de códigos de productos mediante árboles

Incluye:
Punto 1. Árbol Binario de Búsqueda (BST)
Punto 2. Árbol AVL con balanceo y rotaciones
Punto 3. Recorrido Preorden
Punto 4. Recorrido Inorden
Punto 5. Recorrido Postorden
Visualización gráfica comparativa BST vs AVL

Para la visualización se utiliza matplotlib.
Si no lo tienes instalado:
    pip install matplotlib
"""

# ============================================================
# NODO COMÚN PARA LOS DOS ÁRBOLES
# ============================================================

import matplotlib.pyplot as plt

class Nodo:
    def __init__(self, codigo):
        self.codigo = codigo
        self.izq = None
        self.der = None
        self.altura = 1


# ============================================================
# PUNTO 1 - ÁRBOL BINARIO DE BÚSQUEDA (BST)
# ============================================================

class BST:
    def __init__(self):
        self.raiz = None

    def insertar(self, codigo):
        nuevo = Nodo(codigo)

        # Si el árbol está vacío, el nuevo nodo será la raíz.
        if self.raiz is None:
            self.raiz = nuevo
            return

        actual = self.raiz

        # Se inserta respetando:
        # menores a la izquierda y mayores a la derecha.
        while True:
            if codigo < actual.codigo:
                if actual.izq is None:
                    actual.izq = nuevo
                    return
                actual = actual.izq

            elif codigo > actual.codigo:
                if actual.der is None:
                    actual.der = nuevo
                    return
                actual = actual.der

            else:
                # No se insertan códigos repetidos.
                return

    def preorden(self):
        return preorden(self.raiz)

    def inorden(self):
        return inorden(self.raiz)

    def postorden(self):
        return postorden(self.raiz)


# ============================================================
# PUNTO 2 - ÁRBOL AVL
# ============================================================

class AVL:
    def __init__(self):
        self.raiz = None
        self.log_rotaciones = []

    def altura(self, nodo):
        if nodo is None:
            return 0
        return nodo.altura

    def factor_balance(self, nodo):
        """
        Factor de balance:
        altura(izquierda) - altura(derecha)
        """
        if nodo is None:
            return 0

        return self.altura(nodo.izq) - self.altura(nodo.der)

    def rotacion_derecha(self, y):
        x = y.izq
        temporal = x.der

        # Rotación.
        x.der = y
        y.izq = temporal

        # Actualizar alturas.
        y.altura = 1 + max(
            self.altura(y.izq),
            self.altura(y.der)
        )

        x.altura = 1 + max(
            self.altura(x.izq),
            self.altura(x.der)
        )

        return x

    def rotacion_izquierda(self, x):
        y = x.der
        temporal = y.izq

        # Rotación.
        y.izq = x
        x.der = temporal

        # Actualizar alturas.
        x.altura = 1 + max(
            self.altura(x.izq),
            self.altura(x.der)
        )

        y.altura = 1 + max(
            self.altura(y.izq),
            self.altura(y.der)
        )

        return y

    def insertar(self, codigo):
        self.raiz = self._insertar(self.raiz, codigo)

    def _insertar(self, nodo, codigo):

        # ----------------------------------------------------
        # 1. Insertar normalmente como en un BST
        # ----------------------------------------------------
        if nodo is None:
            return Nodo(codigo)

        if codigo < nodo.codigo:
            nodo.izq = self._insertar(nodo.izq, codigo)

        elif codigo > nodo.codigo:
            nodo.der = self._insertar(nodo.der, codigo)

        else:
            # No insertar repetidos.
            return nodo

        # ----------------------------------------------------
        # 2. Actualizar la altura del nodo
        # ----------------------------------------------------
        nodo.altura = 1 + max(
            self.altura(nodo.izq),
            self.altura(nodo.der)
        )

        # ----------------------------------------------------
        # 3. Calcular factor de balance
        # ----------------------------------------------------
        balance = self.factor_balance(nodo)

        # ----------------------------------------------------
        # 4. Revisar los cuatro casos de desbalance
        # ----------------------------------------------------

        # Caso LL: izquierda - izquierda.
        if balance > 1 and codigo < nodo.izq.codigo:
            self.log_rotaciones.append(
                f"Al insertar {codigo}: rotación simple a la DERECHA "
                f"en {nodo.codigo} (caso LL)"
            )
            return self.rotacion_derecha(nodo)

        # Caso RR: derecha - derecha.
        if balance < -1 and codigo > nodo.der.codigo:
            self.log_rotaciones.append(
                f"Al insertar {codigo}: rotación simple a la IZQUIERDA "
                f"en {nodo.codigo} (caso RR)"
            )
            return self.rotacion_izquierda(nodo)

        # Caso LR: izquierda - derecha.
        if balance > 1 and codigo > nodo.izq.codigo:
            self.log_rotaciones.append(
                f"Al insertar {codigo}: rotación DOBLE izquierda-derecha "
                f"en {nodo.codigo} (caso LR)"
            )
            nodo.izq = self.rotacion_izquierda(nodo.izq)
            return self.rotacion_derecha(nodo)

        # Caso RL: derecha - izquierda.
        if balance < -1 and codigo < nodo.der.codigo:
            self.log_rotaciones.append(
                f"Al insertar {codigo}: rotación DOBLE derecha-izquierda "
                f"en {nodo.codigo} (caso RL)"
            )
            nodo.der = self.rotacion_derecha(nodo.der)
            return self.rotacion_izquierda(nodo)

        return nodo

    def preorden(self):
        return preorden(self.raiz)

    def inorden(self):
        return inorden(self.raiz)

    def postorden(self):
        return postorden(self.raiz)


# ============================================================
# PUNTO 3 - RECORRIDO PREORDEN
# Raíz -> Izquierda -> Derecha
# ============================================================

def preorden(nodo, resultado=None):
    if resultado is None:
        resultado = []

    if nodo is not None:
        resultado.append(nodo.codigo)
        preorden(nodo.izq, resultado)
        preorden(nodo.der, resultado)

    return resultado


# ============================================================
# PUNTO 4 - RECORRIDO INORDEN
# Izquierda -> Raíz -> Derecha
# ============================================================

def inorden(nodo, resultado=None):
    if resultado is None:
        resultado = []

    if nodo is not None:
        inorden(nodo.izq, resultado)
        resultado.append(nodo.codigo)
        inorden(nodo.der, resultado)

    return resultado


# ============================================================
# PUNTO 5 - RECORRIDO POSTORDEN
# Izquierda -> Derecha -> Raíz
# ============================================================

def postorden(nodo, resultado=None):
    if resultado is None:
        resultado = []

    if nodo is not None:
        postorden(nodo.izq, resultado)
        postorden(nodo.der, resultado)
        resultado.append(nodo.codigo)

    return resultado


# ============================================================
# CONSTRUIR LOS DOS ÁRBOLES CON LOS MISMOS CÓDIGOS
# ============================================================

def construir_arboles(codigos):
    bst = BST()
    avl = AVL()

    for codigo in codigos:
        bst.insertar(codigo)
        avl.insertar(codigo)

    return bst, avl


# ============================================================
# VISUALIZACIÓN GRÁFICA
# ============================================================

def calcular_posiciones(raiz):
    """
    Calcula una posición distinta para cada nodo.

    La posición horizontal se obtiene siguiendo el recorrido Inorden.
    La posición vertical depende de la profundidad.
    """
    posiciones = {}
    contador_x = [0]

    def recorrer(nodo, profundidad):
        if nodo is None:
            return

        recorrer(nodo.izq, profundidad + 1)

        posiciones[id(nodo)] = (
            contador_x[0],
            -profundidad
        )
        contador_x[0] += 1

        recorrer(nodo.der, profundidad + 1)

    recorrer(raiz, 0)
    return posiciones


def dibujar_subarbol(ax, nodo, posiciones):
    if nodo is None:
        return

    x, y = posiciones[id(nodo)]

    # Línea hacia el hijo izquierdo.
    if nodo.izq is not None:
        x_hijo, y_hijo = posiciones[id(nodo.izq)]

        ax.plot(
            [x, x_hijo],
            [y, y_hijo],
            linewidth=1.5
        )

        dibujar_subarbol(
            ax,
            nodo.izq,
            posiciones
        )

    # Línea hacia el hijo derecho.
    if nodo.der is not None:
        x_hijo, y_hijo = posiciones[id(nodo.der)]

        ax.plot(
            [x, x_hijo],
            [y, y_hijo],
            linewidth=1.5
        )

        dibujar_subarbol(
            ax,
            nodo.der,
            posiciones
        )

    # Dibujar el nodo.
    ax.scatter(
        [x],
        [y],
        s=1300,
        zorder=3
    )

    ax.text(
        x,
        y,
        str(nodo.codigo),
        ha="center",
        va="center",
        fontsize=11,
        zorder=4
    )


def dibujar_arbol(ax, raiz, titulo):
    ax.set_title(
        titulo,
        fontsize=14,
        fontweight="bold"
    )

    ax.axis("off")

    if raiz is None:
        ax.text(
            0.5,
            0.5,
            "Árbol vacío",
            ha="center",
            va="center"
        )
        return

    posiciones = calcular_posiciones(raiz)

    dibujar_subarbol(
        ax,
        raiz,
        posiciones
    )

    xs = [posicion[0] for posicion in posiciones.values()]
    ys = [posicion[1] for posicion in posiciones.values()]

    ax.set_xlim(
        min(xs) - 1,
        max(xs) + 1
    )

    ax.set_ylim(
        min(ys) - 1,
        max(ys) + 1
    )


def mostrar_comparacion(bst, avl):
    """
    Muestra los dos árboles juntos para poder compararlos visualmente.
    """
    figura, ejes = plt.subplots(
        1,
        2,
        figsize=(14, 7)
    )

    dibujar_arbol(
        ejes[0],
        bst.raiz,
        "Árbol Binario de Búsqueda (BST)"
    )

    dibujar_arbol(
        ejes[1],
        avl.raiz,
        "Árbol AVL"
    )

    figura.suptitle(
        "Comparación gráfica: BST vs AVL",
        fontsize=16,
        fontweight="bold"
    )

    plt.tight_layout()
    plt.show()


# ============================================================
# DATOS DE ENTRADA
# ============================================================

def leer_codigos():

    # Pedir cantidad de productos.
    while True:
        try:
            cantidad = int(
                input("Ingrese la cantidad de productos: ")
            )

            if cantidad < 0:
                print(
                    "La cantidad no puede ser negativa."
                )
                continue

            break

        except ValueError:
            print(
                "Ingrese un número entero válido."
            )

    # Pedir cada código.
    codigos = []

    for i in range(cantidad):
        while True:
            try:
                codigo = int(
                    input(
                        f"Ingrese el código {i + 1}: "
                    )
                )

                codigos.append(codigo)
                break

            except ValueError:
                print(
                    "El código debe ser un número entero."
                )

    return codigos


# ============================================================
# PROGRAMA PRINCIPAL
# ============================================================

def main():

    print("=" * 65)
    print(
        "ORGANIZACIÓN DE CÓDIGOS DE PRODUCTOS MEDIANTE ÁRBOLES"
    )
    print("=" * 65)

    # Datos de entrada.
    codigos = leer_codigos()

    print("\nLista original:")
    print(codigos)

    # Construir BST y AVL usando exactamente el mismo orden.
    bst, avl = construir_arboles(codigos)

    # --------------------------------------------------------
    # PUNTO 3 - PREORDEN
    # --------------------------------------------------------
    print(
        "\nPUNTO 3 - PREORDEN "
        "(Raíz -> Izquierda -> Derecha)"
    )

    print(
        "BST:",
        bst.preorden()
    )

    print(
        "AVL:",
        avl.preorden()
    )

    # --------------------------------------------------------
    # PUNTO 4 - INORDEN
    # --------------------------------------------------------
    print(
        "\nPUNTO 4 - INORDEN "
        "(Izquierda -> Raíz -> Derecha)"
    )

    print(
        "BST:",
        bst.inorden()
    )

    print(
        "AVL:",
        avl.inorden()
    )

    print(
        "Característica especial: el recorrido Inorden "
        "de un BST o AVL entrega los códigos ordenados "
        "de menor a mayor."
    )

    # --------------------------------------------------------
    # PUNTO 5 - POSTORDEN
    # --------------------------------------------------------
    print(
        "\nPUNTO 5 - POSTORDEN "
        "(Izquierda -> Derecha -> Raíz)"
    )

    print(
        "BST:",
        bst.postorden()
    )

    print(
        "AVL:",
        avl.postorden()
    )

    # --------------------------------------------------------
    # Mostrar las rotaciones que necesitó el AVL.
    # --------------------------------------------------------
    print(
        "\nROTACIONES REALIZADAS EN EL AVL:"
    )

    if avl.log_rotaciones:
        for rotacion in avl.log_rotaciones:
            print(
                "-",
                rotacion
            )
    else:
        print(
            "No fue necesaria ninguna rotación."
        )

    # --------------------------------------------------------
    # Visualización gráfica.
    # --------------------------------------------------------
    mostrar_comparacion(
        bst,
        avl
    )


if __name__ == "__main__":
    main()
```

```python
# ==========================================================
# 7. TESTS APLICADOS
# ==========================================================

def altura_real_test(nodo):
    if nodo is None:
        return 0

    return 1 + max(
        altura_real_test(nodo.izq),
        altura_real_test(nodo.der)
    )


def factor_balance_test(nodo):
    if nodo is None:
        return 0

    return (
        altura_real_test(nodo.izq)
        -
        altura_real_test(nodo.der)
    )


def arbol_esta_balanceado(nodo):
    if nodo is None:
        return True

    if abs(factor_balance_test(nodo)) > 1:
        return False

    return (
        arbol_esta_balanceado(nodo.izq)
        and
        arbol_esta_balanceado(nodo.der)
    )


def es_bst(
    nodo,
    minimo=float("-inf"),
    maximo=float("inf")
):
    if nodo is None:
        return True

    if not (
        minimo
        <
        nodo.codigo
        <
        maximo
    ):
        return False

    return (
        es_bst(
            nodo.izq,
            minimo,
            nodo.codigo
        )
        and
        es_bst(
            nodo.der,
            nodo.codigo,
            maximo
        )
    )


# ---------- datos del enunciado ----------
codigos_test = [
    30,
    20,
    10,
    25,
    28,
    27,
    5,
    50
]

bst_test, avl_test = construir_arboles(
    codigos_test
)


# ---------- contenido ----------
esperado_ordenado = [
    5,
    10,
    20,
    25,
    27,
    28,
    30,
    50
]

assert inorden(bst_test.raiz) == esperado_ordenado
assert inorden(avl_test.raiz) == esperado_ordenado


# ---------- propiedad BST ----------
assert es_bst(bst_test.raiz) is True
assert es_bst(avl_test.raiz) is True


# ---------- BST sin balancear ----------
assert bst_test.raiz.codigo == 30
assert bst_test.raiz.izq.codigo == 20
assert bst_test.raiz.der.codigo == 50
assert bst_test.raiz.izq.izq.codigo == 10
assert bst_test.raiz.izq.der.codigo == 25

assert preorden(bst_test.raiz) == [
    30, 20, 10, 5, 25, 28, 27, 50
]

assert postorden(bst_test.raiz) == [
    5, 10, 27, 28, 25, 20, 50, 30
]


# ---------- AVL balanceado ----------
assert arbol_esta_balanceado(
    avl_test.raiz
) is True

assert abs(
    factor_balance_test(
        avl_test.raiz
    )
) <= 1

assert avl_test.raiz.codigo == 25

assert preorden(avl_test.raiz) == [
    25, 10, 5, 20, 28, 27, 30, 50
]

assert postorden(avl_test.raiz) == [
    5, 20, 10, 27, 50, 30, 28, 25
]


# ---------- rotación LL ----------
_, avl_ll = construir_arboles(
    [30, 20, 10]
)

assert avl_ll.raiz.codigo == 20
assert avl_ll.raiz.izq.codigo == 10
assert avl_ll.raiz.der.codigo == 30
assert arbol_esta_balanceado(
    avl_ll.raiz
) is True


# ---------- rotación RR ----------
_, avl_rr = construir_arboles(
    [10, 20, 30]
)

assert avl_rr.raiz.codigo == 20
assert avl_rr.raiz.izq.codigo == 10
assert avl_rr.raiz.der.codigo == 30
assert arbol_esta_balanceado(
    avl_rr.raiz
) is True


# ---------- rotación LR ----------
_, avl_lr = construir_arboles(
    [30, 10, 20]
)

assert avl_lr.raiz.codigo == 20
assert avl_lr.raiz.izq.codigo == 10
assert avl_lr.raiz.der.codigo == 30
assert arbol_esta_balanceado(
    avl_lr.raiz
) is True


# ---------- rotación RL ----------
_, avl_rl = construir_arboles(
    [10, 30, 20]
)

assert avl_rl.raiz.codigo == 20
assert avl_rl.raiz.izq.codigo == 10
assert avl_rl.raiz.der.codigo == 30
assert arbol_esta_balanceado(
    avl_rl.raiz
) is True


# ---------- alturas ----------
assert (
    altura_real_test(
        avl_test.raiz
    )
    <=
    altura_real_test(
        bst_test.raiz
    )
)

assert altura_real_test(
    bst_test.raiz
) == 5

assert altura_real_test(
    avl_test.raiz
) == 4


# ---------- un solo nodo ----------
bst_simple, avl_simple = construir_arboles(
    [42]
)

assert preorden(
    bst_simple.raiz
) == [42]

assert inorden(
    bst_simple.raiz
) == [42]

assert postorden(
    bst_simple.raiz
) == [42]

assert preorden(
    avl_simple.raiz
) == [42]

assert inorden(
    avl_simple.raiz
) == [42]

assert postorden(
    avl_simple.raiz
) == [42]


# ---------- árbol vacío ----------
bst_vacio, avl_vacio = construir_arboles(
    []
)

assert bst_vacio.raiz is None
assert avl_vacio.raiz is None

assert preorden(
    bst_vacio.raiz
) == []

assert inorden(
    bst_vacio.raiz
) == []

assert postorden(
    bst_vacio.raiz
) == []

assert preorden(
    avl_vacio.raiz
) == []

assert inorden(
    avl_vacio.raiz
) == []

assert postorden(
    avl_vacio.raiz
) == []


print(
    "Todos los tests pasaron correctamente. ✔"
)
```
